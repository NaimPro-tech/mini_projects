### Graph
- A graph is a non linear data structure that consists a finite set of vertices and a set of edges that connect those vertices. 

### Graph Representation in Python

```python
## Adjacency List

G = {
    "A":["B","C"],
    "B":["D"],
    "C":["D"],
    "D":[],
}
visit_set = set()
```

```python
topo_list = []
def dfs(node, graph, visited, topo_list):
    if node in visited:
        return
    visited.add(node)
    print(node)

    # for node, neighbours in graph.items():
    for neighbour in graph[node]:
        dfs(neighbour, graph, visited, topo_list)
    topo_list.insert(0, node)
    
```

```python
dfs("A",G,visit_set, topo_list=topo_list)
print(topo_list)
```
